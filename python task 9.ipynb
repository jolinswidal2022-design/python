```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("profit_sales_analysis_dataset.csv")
```

```python
df
```

```python
numeric_cols=["Unit_Price","Quantity","Discount_Percent","Sales","Profit","Profit_Margin_Percent","Cost"]
alriya=df[numeric_cols].corr()
alriya
```

```python
#heat map
sns.heatmap(alriya,
           annot=True,
           cmap="coolwarm",
           cbar=False,
           square=True,
           linewidth=1)
plt.title("profit_margin")
plt.show()
```

```python
#bar chart
sns.barplot(x="Category",y="Sales",data=df,width=0.5,color="blue")
plt.title("Sales v/s Category")
plt.show()

```

```python
#line chart
sns.lineplot(x="Category",y="Profit",data=df)
plt.title("Profit v/s Category")
plt.show()
```

```python
#Budget vs Actual charts,
data={
    "department":["HR","Finance","Sales","Marketing","IT","Production","Kannada","Hindi"],
    "Budget":[50000,75000,90000,60000,10000,85000,45000,70000],
    "Actual":[48000,80000,85000,65000,23000,67000,90000,20000]
}

df=pd.DataFrame(data)
print(df)

df_long=df.melt(
    id_vars="department",
    value_vars=["Budget","Actual"],
    var_name="Type",
    value_name="Amount"
)
df_long

sns.barplot(
    data=df_long,
    x="department",
    y="Amount",
    hue="Type"
)
plt.title("Budget vs Actual")
plt.xlabel("department")
plt.ylabel("Amont")
plt.show()
```
