```python
Customer=df.groupby("Customer")[["Sales","Profit"]].sum()
Customer
```

```python
best_customer=Customer["Sales"].idxmax()
print("best customer:",best_customer)
print("Sales:",Customer.max())
```

```python
sales_report=df.groupby("Category").agg(
        Total_sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        average_sales=("Sales","mean"),
        total_quantity=("Quantity","sum")
)
print(sales_report)
```

```python
pivot=pd.pivot_table(
    df,
    values="Sales",
    index="Customer",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)

print(pivot)
```
