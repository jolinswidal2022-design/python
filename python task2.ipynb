```python
transactions = [100, -200, 8000, -500]

for transaction in transactions:
    if transaction > 0:
        print("Credit")
    else:
        print("Debit")
```

```python
transactions = [2000, 8000, 9500, 9000, 2000]

for amount in transactions:
    if amount > 5000:
        print("Large")
    else:
        print("Small")
```

```python
balance = 10000

amount = float(input("Enter debit amount: "))

if amount <= balance:
    print("Debit successful")
else:
    print("Insufficient balance")
```

```python
account = {"balance": 8000}

amount = float(input("Enter amount: "))

if amount <= account["balance"]:
    account["balance"] = account["balance"] - amount
    print("New balance:", account["balance"])
else:
    print("Not enough balance")
```

```python
transactions = [("Credit", 1000), ("Debit", 500)]

for transaction_type, amount in transactions:
    if transaction_type == "Credit":
        print("Added")
    else:
        print("Deducted")
```
