```python
sales_report=df.groupby("Category").agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
)
print(sales_report)
```

```python
customer_summeries=df.groupby(["Customer_ID", "Customer_Name"]).agg(
        Gross_Spend=("Sales", "sum"),
        Net_Profit=("Profit", "sum"),
        Total_Transactions=("Transaction_ID", "count"),
        Avg_Items=("Quantity", "mean"),
)
print(customer_summeries)
```

```python
pd.pivot_table(
    df,
    values="Profit",
    index="City",
    columns="Category",
    aggfunc="sum",
    fill_value=0.0,
    margins=True, 
    margins_name="Total_Profit",
)
```
