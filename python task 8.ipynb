```python
df["curent_ratio"]=df["Current_Assets"]/df["Current_Liabilities"]
df["curent_ratio"]
```

```python
df["ROE"]=(df["Net_Income"]/df["Equity"])*100
```

```python
df["EPS"]=df["Net_Income"]/df["Shares_Outstanding"]
```

```python
plt.figure(figsize=(12,6))
plt.plot(df["Year"],df["EPS"],marker="o")
plt.title("EPS Treand")
plt.xlabel("Year")
plt.ylabel("EPS")
plt.grid
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df["Year"],df["curent_ratio"],marker="o")
plt.title("curent_ratio Treand")
plt.xlabel("Year")
plt.ylabel("curent_ratio")
plt.grid
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df["Year"],df["ROE"],marker="o")
plt.title("ROE Treand")
plt.xlabel("Year")
plt.ylabel("ROE")
plt.grid
plt.show()
```
